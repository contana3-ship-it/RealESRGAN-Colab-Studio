# RealESRGAN x4 — Google Drive → Restore → Google Drive
Bật **GPU T4**, chạy ô bên dưới, xác nhận quyền truy cập Google Drive, chọn video và nhấn **Khôi phục**. Không phải upload, không cần Tunnel.


```python
import os, subprocess, shutil, sys, importlib
repo = "https://github.com/contana3-ship-it/RealESRGAN-Colab-Studio.git"
path = "/content/realesrgan-colab-studio"
if os.path.isdir(os.path.join(path, ".git")):
    subprocess.run(["git", "-C", path, "pull", "--ff-only"], check=True)
else:
    if os.path.exists(path): shutil.rmtree(path)
    subprocess.run(["git", "clone", repo, path], check=True)
subprocess.run(["bash", f"{path}/colab/install_native.sh"], check=True)
if path not in sys.path: sys.path.insert(0, path)
importlib.invalidate_caches()
from colab.drive_app import main
main()
```
